# 📚 Define Libraries

```python
from google.cloud import bigquery
from datetime import datetime

# ---------------------------------------------------------
# 1. Create BigQuery client
# ---------------------------------------------------------
client = bigquery.Client(project="prj-prod-dataplatform")
```

# ❄️ cic_model_cash_v1

```python
print("Starting BigQuery job...")

sq = """ 
create or replace table `dap_ds_poweruser_playground.cic_model_cash_v1` as 
select 
a.* 
, date(lmt.startApplyDateTime) application_date
from 
(
(
WITH parsed as (
select 
safe_cast(customerId as string) customerId, 
digitalLoanAccountId,
case when modelDisplayName like 'Alpha%' then 'cic_model_cash' else modelDisplayName end as modelDisplayName,modelVersionId,
safe_cast(prediction as numeric) prediction,
trenchCategory,
REPLACE(REPLACE(calcFeature, "'", '"'), "None", "null") AS calcFeatures,
    deviceOs osType,
FROM `prj-prod-dataplatform.audit_balance.ml_model_run_details`
where modelDisplayName in ('Alpha-Cash-CIC-Model','Alpha Cash CIC Model','cic_model_cash')
and modelVersionId = 'v1'
)
SELECT
    t.* EXCEPT(calcFeatures),

    -- Numeric features
    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.cic_max_age_all_contracts_snapshot')
        AS FLOAT64
    ) AS cic_max_age_all_contracts_snapshot,

    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.cic_ratio_overdue_contracts_to_granted_contracts')
        AS FLOAT64
    ) AS cic_ratio_overdue_contracts_to_granted_contracts,

    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.cic_ratio_closed_over_new_granted_cnt_24M')
        AS FLOAT64
    ) AS cic_ratio_closed_over_new_granted_cnt_24M,

    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.cic_ratio_risky_contracts_to_granted_contracts')
        AS FLOAT64
    ) AS cic_ratio_risky_contracts_to_granted_contracts,

    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.cic_Short_and_Term_Loans_granted_contracts_cnt_24M')
        AS FLOAT64
    ) AS cic_Short_and_Term_Loans_granted_contracts_cnt_24M,

    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.cic_Personal_Loans_granted_contracts_amt_24M')
        AS FLOAT64
    ) AS cic_Personal_Loans_granted_contracts_amt_24M,

    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.cic_CreditAvgCreditLimit')
        AS FLOAT64
    ) AS cic_CreditAvgCreditLimit,


    -- Categorical / string features
    JSON_VALUE(
        calcFeatures,
        '$.cic_ScoreRange'
    ) AS cic_ScoreRange,

    JSON_VALUE(
        calcFeatures,
        '$.cic_ln_loan_level_user_type'
    ) AS cic_ln_loan_level_user_type,

    JSON_VALUE(
        calcFeatures,
        '$.cic_has_ever_been_overdue'
    ) AS cic_has_ever_been_overdue,

    JSON_VALUE(
        calcFeatures,
        '$.cic_latest_granted_contract_overdue_flag'
    ) AS cic_latest_granted_contract_overdue_flag,

    JSON_VALUE(
        calcFeatures,
        '$.cic_flg_zero_non_granted_ever'
    ) AS cic_flg_zero_non_granted_ever,

    JSON_VALUE(
        calcFeatures,
        '$.cic_flg_zero_granted_ever'
    ) AS cic_flg_zero_granted_ever,
    'Prod' DataSelection

FROM
    parsed AS t
)
union all 
(WITH parsed as (
select 
safe_cast(customerId as string) customerId, 
digitalLoanAccountId,
case when modelDisplayName like 'Alpha%' then 'cic_model_cash' else modelDisplayName end as modelDisplayName,modelVersionId,
safe_cast(prediction as numeric) prediction,
trenchCategory,
REPLACE(REPLACE(calcFeature, "'", '"'), "None", "null") AS calcFeatures,
    deviceOs osType,Data_selection
FROM prj-prod-dataplatform.dap_ds_poweruser_playground.ml_training_model_run_details_20260116
where modelDisplayName in ('Alpha-Cash-CIC-Model','Alpha Cash CIC Model','cic_model_cash')
and modelVersionId = 'v1'
)
SELECT
    t.* EXCEPT(calcFeatures, Data_selection),

    -- Numeric features
    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.cic_max_age_all_contracts_snapshot')
        AS FLOAT64
    ) AS cic_max_age_all_contracts_snapshot,

    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.cic_ratio_overdue_contracts_to_granted_contracts')
        AS FLOAT64
    ) AS cic_ratio_overdue_contracts_to_granted_contracts,

    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.cic_ratio_closed_over_new_granted_cnt_24M')
        AS FLOAT64
    ) AS cic_ratio_closed_over_new_granted_cnt_24M,

    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.cic_ratio_risky_contracts_to_granted_contracts')
        AS FLOAT64
    ) AS cic_ratio_risky_contracts_to_granted_contracts,

    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.cic_Short_and_Term_Loans_granted_contracts_cnt_24M')
        AS FLOAT64
    ) AS cic_Short_and_Term_Loans_granted_contracts_cnt_24M,

    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.cic_Personal_Loans_granted_contracts_amt_24M')
        AS FLOAT64
    ) AS cic_Personal_Loans_granted_contracts_amt_24M,

    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.cic_CreditAvgCreditLimit')
        AS FLOAT64
    ) AS cic_CreditAvgCreditLimit,


    -- Categorical / string features
    JSON_VALUE(
        calcFeatures,
        '$.cic_ScoreRange'
    ) AS cic_ScoreRange,

    JSON_VALUE(
        calcFeatures,
        '$.cic_ln_loan_level_user_type'
    ) AS cic_ln_loan_level_user_type,

    JSON_VALUE(
        calcFeatures,
        '$.cic_has_ever_been_overdue'
    ) AS cic_has_ever_been_overdue,

    JSON_VALUE(
        calcFeatures,
        '$.cic_latest_granted_contract_overdue_flag'
    ) AS cic_latest_granted_contract_overdue_flag,

    JSON_VALUE(
        calcFeatures,
        '$.cic_flg_zero_non_granted_ever'
    ) AS cic_flg_zero_non_granted_ever,

    JSON_VALUE(
        calcFeatures,
        '$.cic_flg_zero_granted_ever'
    ) AS cic_flg_zero_granted_ever,
    Data_selection as DataSelection

FROM
    parsed AS t
)
) as a
left join (select digitalLoanAccountId, startApplyDateTime from `risk_credit_mis.loan_master_table`) lmt on lmt.digitalLoanAccountId = a.digitalLoanAccountId
;
    

"""

query_job = client.query(sq)
# Wait until query finishes
query_job.result()
print("BigQuery job completed successfully!")


# ---------------------------------------------------------
# 3. Job details
# ---------------------------------------------------------
print("\n========== JOB DETAILS ==========")

print(f"Job ID              : {query_job.job_id}")
print(f"State               : {query_job.state}")
print(f"Created             : {query_job.created}")
print(f"Started             : {query_job.started}")
print(f"Ended               : {query_job.ended}")

# Bytes processed
if query_job.total_bytes_processed is not None:
    print(
        f"Bytes processed     : "
        f"{query_job.total_bytes_processed:,} "
        f"({query_job.total_bytes_processed / 1024**3:.3f} GB)"
    )

if query_job.total_bytes_billed is not None:
    print(
        f"Bytes billed        : "
        f"{query_job.total_bytes_billed:,} "
        f"({query_job.total_bytes_billed / 1024**3:.3f} GB)"
    )


# ---------------------------------------------------------
# 4. Get newly created table
# ---------------------------------------------------------
table_id = (
    "prj-prod-dataplatform."
    "dap_ds_poweruser_playground."
    "cic_model_cash_v1"
)

table = client.get_table(table_id)


# ---------------------------------------------------------
# 5. Table shape and metadata
# ---------------------------------------------------------
rows = table.num_rows
columns = len(table.schema)

print("\n========== TABLE DETAILS ==========")

print(f"Table               : {table_id}")
print(f"Shape               : ({rows:,}, {columns})")
print(f"Rows                : {rows:,}")
print(f"Columns             : {columns:,}")

print(
    f"Table size          : "
    f"{table.num_bytes:,} bytes "
    f"({table.num_bytes / 1024**2:.2f} MB)"
)

print(f"Created             : {table.created}")
print(f"Last modified       : {table.modified}")


# ---------------------------------------------------------
# 6. Print schema / column details
# ---------------------------------------------------------
print("\n========== SCHEMA ==========")

for i, field in enumerate(table.schema, start=1):
    print(
        f"{i:02d}. "
        f"{field.name:<30} "
        f"{field.field_type:<15} "
        f"{field.mode}"
    )
```
