# 📚 Define Libraries

```python
from google.cloud import bigquery
from datetime import datetime

# ---------------------------------------------------------
# 1. Create BigQuery client
# ---------------------------------------------------------
client = bigquery.Client(project="prj-prod-dataplatform")
```

# ❄️ beta_events_model_cash_v1

```python
print("Starting BigQuery job...")

sq = """ 
create or replace table `dap_ds_poweruser_playground.beta_events_model_cash_v1` as 
select a.* 
, date(lmt.startApplyDateTime) application_date
from 
(
(WITH parsed as (
select 
safe_cast(customerId as string) customerId, 
digitalLoanAccountId,
modelDisplayName,
modelVersionId,
safe_cast(prediction as numeric) prediction,
trenchCategory,
REPLACE(REPLACE(calcFeature, "'", '"'), "None", "null") AS calcFeatures,
deviceOs osType, start_time,
FROM `prj-prod-dataplatform.audit_balance.ml_model_run_details`
where modelDisplayName in  ('beta_events_model_cash')
and modelVersionId like '%v1%'
)
,
latest_request as (
select * from parsed
QUALIFY ROW_NUMBER() OVER (PARTITION BY customerId, digitalLoanAccountId,modelDisplayName ORDER BY start_time DESC ) = 1)
SELECT
    t.* EXCEPT(calcFeatures, start_time),

    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.total_events_cnt_7d')
        AS FLOAT64
    ) AS total_events_cnt_7d,

    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.active_days_cnt_7d')
        AS FLOAT64
    ) AS active_days_cnt_7d,

    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.event_gap_avg_sec_7d')
        AS FLOAT64
    ) AS event_gap_avg_sec_7d,

    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.journey_active_total_mins_7d')
        AS FLOAT64
    ) AS journey_active_total_mins_7d,

    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.loan_calc_total_ratio_7d')
        AS FLOAT64
    ) AS loan_calc_total_ratio_7d,

    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.calculator_repeat_cnt_7d')
        AS FLOAT64
    ) AS calculator_repeat_cnt_7d,

    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.calculator_time_spent_avg_7d')
        AS FLOAT64
    ) AS calculator_time_spent_avg_7d,

    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.summary_time_spent_avg_7d')
        AS FLOAT64
    ) AS summary_time_spent_avg_7d,

    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.summary_with_vas_cnt_7d')
        AS FLOAT64
    ) AS summary_with_vas_cnt_7d,

    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.vas_during_journey_cnt_7d')
        AS FLOAT64
    ) AS vas_during_journey_cnt_7d,

    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.vas_info_cnt_7d')
        AS FLOAT64
    ) AS vas_info_cnt_7d,

    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.vas_to_income_sec_7d')
        AS FLOAT64
    ) AS vas_to_income_sec_7d,

    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.popup_no_vas_cnt_7d')
        AS FLOAT64
    ) AS popup_no_vas_cnt_7d,

    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.dropoff_before_ref2_cnt_7d')
        AS FLOAT64
    ) AS dropoff_before_ref2_cnt_7d,

    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.purpose_to_income_log_sec_7d')
        AS FLOAT64
    ) AS purpose_to_income_log_sec_7d,

    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.purpose_to_income_med_mins_7d')
        AS FLOAT64
    ) AS purpose_to_income_med_mins_7d,

    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.time_on_ref1_max_sec_7d')
        AS FLOAT64
    ) AS time_on_ref1_max_sec_7d,

    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.tod_variance_7d')
        AS FLOAT64
    ) AS tod_variance_7d,

    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.days_on_book')
        AS FLOAT64
    ) AS days_on_book,

    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.event_score')
        AS FLOAT64
    ) AS event_score,
    'Prod' DataSelection

FROM latest_request AS t
)
union all 

(WITH parsed as (
select 
safe_cast(customerId as string) customerId, 
digitalLoanAccountId,
modelDisplayName,
modelVersionId,
safe_cast(prediction as numeric) prediction,
trenchCategory,
REPLACE(REPLACE(calcFeature, "'", '"'), "None", "null") AS calcFeatures,
deviceOs osType, Data_selection
FROM prj-prod-dataplatform.dap_ds_poweruser_playground.ml_training_model_run_details_20260116
where modelDisplayName in  ('beta_events_model_cash')
and modelVersionId like '%v1%'
)
SELECT
    t.* EXCEPT(calcFeatures, Data_selection),

    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.total_events_cnt_7d')
        AS FLOAT64
    ) AS total_events_cnt_7d,

    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.active_days_cnt_7d')
        AS FLOAT64
    ) AS active_days_cnt_7d,

    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.event_gap_avg_sec_7d')
        AS FLOAT64
    ) AS event_gap_avg_sec_7d,

    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.journey_active_total_mins_7d')
        AS FLOAT64
    ) AS journey_active_total_mins_7d,

    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.loan_calc_total_ratio_7d')
        AS FLOAT64
    ) AS loan_calc_total_ratio_7d,

    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.calculator_repeat_cnt_7d')
        AS FLOAT64
    ) AS calculator_repeat_cnt_7d,

    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.calculator_time_spent_avg_7d')
        AS FLOAT64
    ) AS calculator_time_spent_avg_7d,

    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.summary_time_spent_avg_7d')
        AS FLOAT64
    ) AS summary_time_spent_avg_7d,

    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.summary_with_vas_cnt_7d')
        AS FLOAT64
    ) AS summary_with_vas_cnt_7d,

    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.vas_during_journey_cnt_7d')
        AS FLOAT64
    ) AS vas_during_journey_cnt_7d,

    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.vas_info_cnt_7d')
        AS FLOAT64
    ) AS vas_info_cnt_7d,

    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.vas_to_income_sec_7d')
        AS FLOAT64
    ) AS vas_to_income_sec_7d,

    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.popup_no_vas_cnt_7d')
        AS FLOAT64
    ) AS popup_no_vas_cnt_7d,

    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.dropoff_before_ref2_cnt_7d')
        AS FLOAT64
    ) AS dropoff_before_ref2_cnt_7d,

    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.purpose_to_income_log_sec_7d')
        AS FLOAT64
    ) AS purpose_to_income_log_sec_7d,

    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.purpose_to_income_med_mins_7d')
        AS FLOAT64
    ) AS purpose_to_income_med_mins_7d,

    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.time_on_ref1_max_sec_7d')
        AS FLOAT64
    ) AS time_on_ref1_max_sec_7d,

    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.tod_variance_7d')
        AS FLOAT64
    ) AS tod_variance_7d,

    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.days_on_book')
        AS FLOAT64
    ) AS days_on_book,

    SAFE_CAST(
        JSON_VALUE(calcFeatures, '$.event_score')
        AS FLOAT64
    ) AS event_score,
    Data_selection DataSelection
FROM parsed AS t
)
)a
left join (select digitalLoanAccountId, startApplyDateTime from `risk_credit_mis.loan_master_table`) lmt on lmt.digitalLoanAccountId = a.digitalLoanAccountId
;
"""

query_job = client.query(sq)
# Wait until query finishes
query_job.result()
print("BigQuery job completed successfully!")


# ---------------------------------------------------------
# 3. Job details
# ---------------------------------------------------------
print("\n========== JOB DETAILS ==========")

print(f"Job ID              : {query_job.job_id}")
print(f"State               : {query_job.state}")
print(f"Created             : {query_job.created}")
print(f"Started             : {query_job.started}")
print(f"Ended               : {query_job.ended}")

# Bytes processed
if query_job.total_bytes_processed is not None:
    print(
        f"Bytes processed     : "
        f"{query_job.total_bytes_processed:,} "
        f"({query_job.total_bytes_processed / 1024**3:.3f} GB)"
    )

if query_job.total_bytes_billed is not None:
    print(
        f"Bytes billed        : "
        f"{query_job.total_bytes_billed:,} "
        f"({query_job.total_bytes_billed / 1024**3:.3f} GB)"
    )


# ---------------------------------------------------------
# 4. Get newly created table
# ---------------------------------------------------------
table_id = (
    "prj-prod-dataplatform."
    "dap_ds_poweruser_playground."
    "beta_events_model_cash_v1"
)

table = client.get_table(table_id)


# ---------------------------------------------------------
# 5. Table shape and metadata
# ---------------------------------------------------------
rows = table.num_rows
columns = len(table.schema)

print("\n========== TABLE DETAILS ==========")

print(f"Table               : {table_id}")
print(f"Shape               : ({rows:,}, {columns})")
print(f"Rows                : {rows:,}")
print(f"Columns             : {columns:,}")

print(
    f"Table size          : "
    f"{table.num_bytes:,} bytes "
    f"({table.num_bytes / 1024**2:.2f} MB)"
)

print(f"Created             : {table.created}")
print(f"Last modified       : {table.modified}")


# ---------------------------------------------------------
# 6. Print schema / column details
# ---------------------------------------------------------
print("\n========== SCHEMA ==========")

for i, field in enumerate(table.schema, start=1):
    print(
        f"{i:02d}. "
        f"{field.name:<30} "
        f"{field.field_type:<15} "
        f"{field.mode}"
    )
```
